```python
import numpy as np
import math
```

```python
words = open("names.txt").read().splitlines()
```

```python
words[:10]
```

```python
chars = "abcdefghijklmnopqrstuvwxyz"
stoi = {c: i for i, c in enumerate(chars)}      # char -> index, a=0 b=1 ...
stoi["."] = 26                                  # add boundary token
itos = {i: c for c, i in stoi.items()}          # reverse mapping
```

```python
stoi["c"]
```

```python
N = np.zeros((27, 27), dtype=np.int32)
```

```python
for w in words:
    padded = "." + w + "."
    for c1, c2 in zip(padded, padded[1:]):          # consecutive pairs
        N[stoi[c1], stoi[c2]] += 1        
```

```python
N[3]
```

```python
row_sums = N.sum(axis=1, keepdims=True)
```

```python
P = (N+1)/(row_sums+27)
```

```python
logP = np.log(P)
```

```python
NLL = -1 * (N * logP).sum() / N.sum()
```

```python
NLL
```

```python
perplexity = math.exp(NLL)
print(perplexity)
```

```python
total_names = []
while len(total_names) < 10:
    name = ""
    ix = 26
    curr_char = ''
    while curr_char != ".":
        ix = np.random.choice(27, p=P[ix])
        curr_char = itos[ix]
        name += curr_char
    total_names.append(name)
    print(name)
```
