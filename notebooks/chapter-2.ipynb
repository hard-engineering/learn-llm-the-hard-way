```python
import numpy as np
import math
```

```python
words = open("names.txt").read().splitlines()
```

```python
words[:10]
```

```python
chars = "abcdefghijklmnopqrstuvwxyz"
stoi = {c: i for i, c in enumerate(chars)}      # char -> index, a=0 b=1 ...
stoi["."] = 26                                  # add boundary token
itos = {i: c for c, i in stoi.items()}          # reverse mapping
```

```python
# Random weights
W = np.random.randn(27, 27)
```

```python
xs = []
ys = []
n = 0
for w in words:
    padded = "." + w + "."
    for c1, c2 in zip(padded, padded[1:]):          # consecutive pairs
        xs.append(stoi[c1])
        ys.append(stoi[c2])
        n += 1
```

```python
X = np.eye(27)[xs]
Y = np.eye(27)[ys]
rate = 50
```

```python
for i in range(200):
    logits = X @ W
    exp = np.exp(logits)
    row_sums = exp.sum(axis=1, keepdims=True)
    probs = exp/row_sums
    loss = -np.log(probs[np.arange(n), ys]).mean()
    if i % 5 == 0:
        print(i, loss)
    dlogits = probs.copy()
    dlogits[np.arange(n), ys] -= 1
    dlogits /= n
    dW = X.T @ dlogits
    W = W - rate * dW
```
